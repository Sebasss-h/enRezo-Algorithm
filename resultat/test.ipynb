```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
time_df = pd.read_csv('time.csv', sep=' ')
```

```python
time_df
```

```python
fig = plt.figure()
fig.subplots_adjust(hspace=0.4, wspace=0.4)
ax = fig.add_subplot(2, 2, 1)
sns.scatterplot(data=time_df, x="NbBats", y="Proj", ax=ax)
ax = fig.add_subplot(2, 2, 2)
sns.scatterplot(data=time_df, x="NbBats", y="Graph", ax=ax)
ax = fig.add_subplot(2, 2, 3)
sns.scatterplot(data=time_df, x="NbBats", y="Tree", ax=ax)
ax = fig.add_subplot(2, 2, 4)
sns.scatterplot(data=time_df, x="NbBats", y="Trace", ax=ax)
plt.show()
```
